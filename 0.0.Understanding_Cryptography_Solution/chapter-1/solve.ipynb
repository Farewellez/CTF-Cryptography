### 1.8. What is the multiplicative inverse of 5 in Z11, Z12, and Z13? You can do a trial-and-error search using a calculator or a PC.

With this simple problem we want now to stress the fact that the inverse of an
integer in a given ring depends completely on the ring considered. That is, if the
modulus changes, the inverse changes. Hence, it doesn’t make sense to talk about
an inverse of an element unless it is clear what the modulus is. This fact is crucial
for the RSA cryptosystem, which is introduced in Chap. 7. The extended Euclidean
algorithm, which can be used for computing inverses efficiently, is introduced in
Sect. 6.3.

````python
from math import gcd

z11 = pow(5,-1,11)
assert gcd(5,11) == 1

z12 = pow(5,-1,12)
assert gcd(5,12) == 1

z13 = pow(5,-1,13)
assert gcd(5,13) == 1

print(f"z11 = {z11}")
print(f"z12 = {z12}")
print(f"z13 = {z13}")
````

#### 1.10. Find all integers n between 0 ≤ n < m that are relatively prime to m for m = 4,5,9,26. We denote the number of integers n which fulfill the condition by φ(m), e.g. φ(3)= 2. This function is called “Euler’s phi function”. What is φ(m) for m = 4,5,9,26?

````python
from math import gcd
m = [4,5,9,26]
for mod in m:
    φ = []
    for n in range(0,mod):
        if gcd(n,mod) == 1:
            φ.append(n)
    print(f"φ({mod}) = {len(φ)}")
       
````

#### number 3
```
For verification: (1) encrypt the letter K first with ek1 and the result with ek2, and
(2) encrypt the letter K with ek3.
```

````python
from string import ascii_uppercase

def enc1(x):
    a1 = 3
    b1 = 5
    y = ((a1 * ascii_uppercase.index(x)) + b1) % 26
    return ascii_uppercase[y]

def enc2(x):
    a2 = 11
    b2 = 7
    y = ((a2 * ascii_uppercase.index(x)) + b2) % 26
    return ascii_uppercase[y]

def enc3(x):
    a3 = 7
    b3 = 10
    y = ((a3 * ascii_uppercase.index(x)) + b3) % 26
    return ascii_uppercase[y]

c1 = enc1('K')
c2 = enc2(c1)
c3 = enc3('K')

print(f"Hasil ek2(ek1('K')): {c2}")
print(f"Hasil ek3('K'): {c3}")
````
