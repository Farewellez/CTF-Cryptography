```python
from string import ascii_lowercase, ascii_uppercase
```

```python
# the ROT13 is subtitution cipher that replace p[i] -> [i+13]
ct = "cvpbPGS{abg_gbb_onq_bs_n_ceboyrz}"
pt = ""

for c in ct:
    if c in ascii_lowercase:
        sub = (ascii_lowercase.index(c) + 13) % 26
        pt += ascii_lowercase[sub]
    elif c in ascii_uppercase:
        sub = (ascii_uppercase.index(c) + 13) % 26
        pt += ascii_uppercase[sub]
    else:
        pt += c

print(pt)
```
